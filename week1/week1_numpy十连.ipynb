```python
import numpy as np


a = np.arange(20).reshape(4,5)
print(a)
print(np.zeros((2,3)))
print(a[0],a[:,2],a[1:,::2])
print(a[a > 10])
print((a+10).shape,(a*2)[0,0])
x = np.arange(4).reshape(4,1); y = np.arange(3).reshape(1,3)
print((x+y).shape)
b = np.random.randn(4,5)
print((a*b)[0,0],(a @ b.T)[0,0])
print(a.sum(),a.sum(axis=0))
print(a.mean(),a.mean(axis=1))
print(a.T.shape,a.reshape(2, 10).shape)
```
